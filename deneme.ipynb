```python
import yfinance as yf
import time
from IPython.display import clear_output
from datetime import datetime

# Create ticker objects
gold = yf.Ticker("GC=F")
silver = yf.Ticker("SI=F")

print("Data stream is starting... (To stop, click the 'Square' or 'Stop' button next to the cell)")
time.sleep(2)

try:
    while True:
        # Clear old notebook output (prevents scrolling)
        clear_output(wait=True)
        
        # Fetch prices
        gold_price = gold.fast_info['last_price']
        silver_price = silver.fast_info['last_price']
        
        # Calculate the ratio
        ratio = gold_price / silver_price
        now = datetime.now().strftime("%H:%M:%S")
        
        # Print nicely to the screen
        print(f"⏱️ Last Update: {now}")
        print("-" * 35)
        print(f"🥇 Gold (GC=F)        : ${gold_price:.2f}")
        print(f"🥈 Silver (SI=F)        : ${silver_price:.2f}")
        print(f"📊 Gold/Silver Ratio : {ratio:.2f}")
        print("-" * 35)
        
        # Simple alert logic
        if ratio > 80.0:
            print("🚨 ALERT: Ratio is above 80! (Silver is cheap)")
        elif ratio < 75.0:
            print("⚠️ INFO: Ratio is below 75! (Silver is expensive)")
        else:
            print("✅ Market is stable right now.")
            
        # Wait 5 seconds to avoid overloading Yahoo
        time.sleep(5)
        
except KeyboardInterrupt:
    clear_output(wait=True)
    print("🛑 Data stream stopped by you.")
```

```python
import yfinance as yf
import time
from datetime import datetime

# Create all ticker objects
gold = yf.Ticker("GC=F")
silver = yf.Ticker("SI=F")
btc = yf.Ticker("BTC-USD")
dxy = yf.Ticker("DX-Y.NYB")
eur_usd = yf.Ticker("EURUSD=X")

print("🚀 Enriched data stream started...")
print("To stop, click the 'Square' or 'Stop' button next to the cell.")
print("-" * 80)
time.sleep(2)

try:
    while True:
        # Fetch prices
        gold_price = gold.fast_info['last_price']
        silver_price = silver.fast_info['last_price']
        btc_price = btc.fast_info['last_price']
        dxy_price = dxy.fast_info['last_price']
        eur_price = eur_usd.fast_info['last_price']
        
        # Compute synthetic metrics on the fly
        gs_ratio = gold_price / silver_price
        gold_in_eur = gold_price / eur_price 
        
        now = datetime.now().strftime("%H:%M:%S")
        
        # Clean log format for continuous output
        log_mesaji = (
            f"[{now}] "
            f"🥇 Gold: ${gold_price:,.2f} | 🥈 Silver: ${silver_price:.2f} | "
            f"⚖️ Ratio: {gs_ratio:.2f} | "
            f"₿ BTC: ${btc_price:,.0f} | 💵 DXY: {dxy_price:.2f} | 🇪🇺 Gold(€): €{gold_in_eur:,.2f}"
        )
        print(log_mesaji)
        
        # Dynamic alert mechanism (prints extra line only on alarms)
        if gs_ratio > 80.0:
            print(f"   >>> 🚨 ALERT: Gold/Silver ratio exceeded 80! Silver is cheap.")
        elif gs_ratio < 75.0:
            print(f"   >>> ⚠️ INFO: Ratio is below 75! Silver got expensive.")
            
        # Wait 5 seconds to avoid overloading Yahoo Finance
        time.sleep(5)
        
except KeyboardInterrupt:
    print("\n🛑 Data stream successfully stopped by you.")
```

```python
import yfinance as yf
import time
from datetime import datetime
import warnings

# Hide warnings to keep output clean
warnings.filterwarnings('ignore')

print("🚀 Data stream started (single-fetch method)...")
print("To stop, press the 'Stop' button.")
print("-" * 80)
time.sleep(2)

try:
    while True:
        # Define ticker objects INSIDE the loop.
        # This clears old cache each loop and fetches freshest data.
        # Even when markets are closed, returns last known price instead of NaN.
        gold_price = yf.Ticker("GC=F").fast_info['last_price']
        silver_price = yf.Ticker("SI=F").fast_info['last_price']
        btc_price = yf.Ticker("BTC-USD").fast_info['last_price']
        dxy_price = yf.Ticker("DX-Y.NYB").fast_info['last_price']
        eur_price = yf.Ticker("EURUSD=X").fast_info['last_price']
        
        # Compute synthetic metrics on the fly
        gs_ratio = gold_price / silver_price
        gold_in_eur = gold_price / eur_price 
        
        now = datetime.now().strftime("%H:%M:%S")
        
        # Clean log format for continuous output
        log_mesaji = (
            f"[{now}] "
            f"🥇 Gold: ${gold_price:,.2f} | 🥈 Silver: ${silver_price:.2f} | "
            f"⚖️ Ratio: {gs_ratio:.2f} | "
            f"₿ BTC: ${btc_price:,.0f} | 💵 DXY: {dxy_price:.2f} | 🇪🇺 Gold(€): €{gold_in_eur:,.2f}"
        )
        print(log_mesaji)
        
        # Dynamic alert mechanism
        if gs_ratio > 80.0:
            print(f"   >>> 🚨 ALERT: Gold/Silver ratio exceeded 80! Silver is cheap.")
        elif gs_ratio < 75.0:
            print(f"   >>> ⚠️ INFO: Ratio is below 75! Silver got expensive.")
            
        # Wait 5 seconds to avoid overloading Yahoo Finance
        time.sleep(5)
        
except KeyboardInterrupt:
    print("\n🛑 Data stream successfully stopped by you.")
except Exception as e:
    print(f"\n❌ An error occurred: {e}")
```
